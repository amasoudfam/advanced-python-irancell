```python
def getPrime():
    def isPrime(n):
        if n == 2:
            return True
        if n % 2 == 0:
            return False
        for i in range(3, int(n**0.5) + 1, 2):
            if n % i == 0:
                return False
        return True
    m = 2
    while True:
        while not isPrime(m):
            m += 1
        m+=1
        yield m-1
```

```python
def read_file(filename, n_line):
    with open(filename) as f:
        while True:
            result = []
            for line in f:
                result.append(line)
                if len(result) == n_line:
                    break
            if len(result) == 0:
                break
            yield result
for x in read_file('test.txt', 10):
    print(x)
```

```python
def calc_time(func):
    def wrapper(*arg, **args):
        print('Before run')
        func(*arg, **args)
        print('After run')
    return wrapper
    
@calc_time
def test_func(message):
    print('Hello', message)

test_func('Python')
```

```python
import time

def calc_time(func):
    def wrapper(*arg, **args):
        start_time = time.time()
        func(*arg, **args)
        end_time = time.time()
        print(f'{func.__name__} exec time:', end_time - start_time)
    return wrapper
    
@calc_time
def test_func(message):
    print('Hello', message)
    time.sleep(1)

test_func('Python')
```
