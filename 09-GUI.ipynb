```python
import tkinter as tk

def calculate():
    try:
        value = eval(entry.get())
        result.set(value)
    except:
        result.set("خطا!")

root = tk.Tk()
root.title("یک مثال دیگر")
root.resizable(False, False)

result = tk.StringVar()

entry = tk.Entry(root, textvariable=result, font=("Arial", 14), width=20)
entry.pack(pady=10)

button = tk.Button(root, text="برو!", command=calculate, font=("Arial", 12))
button.pack(pady=5)

root.mainloop()
```

```python
import tkinter as tk
import random

def play(choice):
    options = ["سنگ", "کاغذ", "قیچی"]
    computer_choice = random.choice(options)
    
    if choice == computer_choice:
        result.set("مساوی شد!")
    elif (choice == "سنگ" and computer_choice == "قیچی") or \
         (choice == "کاغذ" and computer_choice == "سنگ") or \
         (choice == "قیچی" and computer_choice == "کاغذ"):
        result.set("شما بردید! ")
    else:
        result.set("کامپیوتر برد!")

root = tk.Tk()
root.resizable(False, False)
root.title("سنگ، کاغذ، قیچی")

result = tk.StringVar()
result.set("انتخاب کن!")

label = tk.Label(root, textvariable=result, font=("Arial", 14))
label.pack(pady=10)

for choice in ["سنگ", "کاغذ", "قیچی"]:
    button = tk.Button(root, text=choice, command=lambda ch=choice: play(ch), font=("Arial", 12))
    button.pack(side=tk.LEFT, padx=10, pady=10)

root.mainloop()
```

```python
import tkinter as tk
from tkinter import END

def on_click(event):
    text = event.widget.cget("text")
    input_text = str(entry.get())
    if text == "=":
        try:
            result = eval(input_text)
            entry.delete(0, END)
            entry.insert(END, result)
        except Exception as e:
            entry.delete(0, END)
            entry.insert(END, "Error")
    elif text == "C":
        entry.delete(0, END)
        entry.insert(END, '0')
    else:
        if input_text == '0':
            entry.delete(0, END)
        if text in '+-*/':
            entry.delete(0, END)
            input_text = eval(input_text)
            entry.insert(END, input_text)
        entry.insert(END, text)

root = tk.Tk()
root.title("Calculator")
root.geometry("350x400")
root.resizable(False, False)

entry = tk.Entry(root, font=("Tahoma", 20), justify='right')
entry.insert(END, '0')
entry.pack()

button_frame = tk.Frame(root)
button_frame.pack()

buttons = [
    ['7', '8', '9', '/'],
    ['4', '5', '6', '*'],
    ['1', '2', '3', '-'],
    ['C', '0', '=', '+']
]

for row in buttons:
    frame = tk.Frame(button_frame)
    frame.pack()
    for button_txt in row:
        btn = tk.Button(frame, text=button_txt, font=("Tahoma", 20), width=5, height=2)
        btn.pack(side=tk.LEFT)
        btn.bind("<Button-1>", on_click)

root.mainloop()
```
