```python
class Circle:
    def __init__(self, r):
        self.setRadius(r)
        
    def setRadius(self, r):
        if r <= 0:
            raise ValueError('Invalid radius')
        self.radius = r

    def getArea(self):
        return self.radius ** 2 * 3.1415

    def getPerimeter(self):
        return self.radius * 2 * 3.1415
```

```python
c = Circle(5)
print(c.radius)
```

```python
c = Circle(-5)
print(c.radius)
```

```python
class Circle:
    def __init__(self, r):
        self.setRadius(r)
        
    def setRadius(self, r):
        if r <= 0:
            raise ValueError('Invalid radius')
        self.__radius = r

    def getArea(self):
        return self.__radius ** 2 * 3.1415

    def getPerimeter(self):
        return self.__radius * 2 * 3.1415
```

```python
c=Circle(5)
```

```python
from abc import ABC, abstractmethod

class Animal(ABC):
    @abstractmethod
    def make_sound(self):
        pass  # این متد فقط تعریف شده و پیاده‌سازی ندارد

class Dog(Animal):
    def make_sound(self):
        return "Woof! Woof!"

class Cat(Animal):
    def make_sound(self):
        return "Meow!"

# animal = Animal()

dog = Dog()
cat = Cat()

print(dog.make_sound())  # Woof! Woof!
print(cat.make_sound())  # Meow!
```
