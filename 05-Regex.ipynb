```python
from selenium import webdriver
from bs4 import BeautifulSoup
classes = 'col-xs-6 col-sm-6 col-md-3 col-lg-3 full_size'
url = 'https://shahrvand.ir/fa/product/cat/%D8%A2%D8%B1%D8%A7%DB%8C%D8%B4%DB%8C-%D8%A8%D9%87%D8%AF%D8%A7%D8%B4%D8%AA%DB%8C/%D8%AF%D8%B3%D8%AA%D9%85%D8%A7%D9%84-%DA%A9%D8%A7%D8%BA%D8%B0%DB%8C-%D9%88-%D8%B3%D9%84%D9%88%D9%84%D8%B2%DB%8C/250-%D8%AF%D8%B3%D8%AA%D9%85%D8%A7%D9%84-%DA%A9%D8%A7%D8%BA%D8%B0%DB%8C.html#/pagesize-1000/stock-1'
driver = webdriver.Edge()
driver.get(url)
driver.implicitly_wait(5)
content = driver.page_source
driver.quit()
soup = BeautifulSoup(content, 'html.parser')
names = []
prices = []
for section in soup.find_all('div', attrs={'class': classes}):
    names.append(section.find('a', attrs={'class': 'title'}).text)
    prices.append(section.find('div', attrs={'class': 'new_price'}).text.strip())
```

```python
import pandas as pd
df = pd.DataFrame({'title': names, 'price': prices})
```
