```python
import pygame
import random

pygame.init()
screen = pygame.display.set_mode((500, 400))
pygame.display.set_caption("یک مثال")

circle_x, circle_y = 100, 100
circle_radius = 100
circle_color = (0, 0, 255)

running = True
clock = pygame.time.Clock()
step = 5
while running:
    clock.tick(60)
    for event in pygame.event.get():
        if event.type == pygame.QUIT:
            running = False
            break
        if event.type == pygame.MOUSEBUTTONDOWN:
            circle_color = (random.randint(0, 255), random.randint(0, 255), random.randint(0, 255))
    keys = pygame.key.get_pressed()
    if keys[pygame.K_LEFT]:
        circle_x -= step
    if keys[pygame.K_RIGHT]:
        circle_x += step
    if keys[pygame.K_UP]:
        circle_y -= step
    if keys[pygame.K_DOWN]:
        circle_y += step
    screen.fill((255, 255, 255))
    pygame.draw.circle(screen, circle_color, (circle_x, circle_y), circle_radius)
    pygame.display.update()
pygame.quit()
```

```python
import pygame
import random

pygame.init()
screen = pygame.display.set_mode((500, 400))
pygame.display.set_caption("یک مثال")

circle_x, circle_y = 100, 100
circle_radius = 40
circle_color = (0, 0, 255)

running = True
clock = pygame.time.Clock()
speed_x = speed_y = 10
while running:
    clock.tick(60)
    for event in pygame.event.get():
        if event.type == pygame.QUIT:
            running = False
            break
        if event.type == pygame.MOUSEBUTTONDOWN:
            circle_color = (random.randint(0, 255), random.randint(0, 255), random.randint(0, 255))
    circle_x += speed_x
    circle_y += speed_y
    if circle_x - circle_radius <= 0 or circle_x + circle_radius >= 500:
        speed_x = -speed_x
    if circle_y - circle_radius <= 0 or circle_y + circle_radius >= 400:
        speed_y = -speed_y
    screen.fill((255, 255, 255))
    pygame.draw.circle(screen, circle_color, (circle_x, circle_y), circle_radius)
    pygame.display.update()
pygame.quit()
```

```python
import pygame
import random

pygame.init()

WIDTH, HEIGHT = 600, 600
screen = pygame.display.set_mode((WIDTH, HEIGHT))
pygame.display.set_caption("بازی مارپله")

board_img = pygame.image.load("ladder.jpeg")  
board_img = pygame.transform.scale(board_img, (WIDTH, HEIGHT))  

player_radius = 15  
player_pos = 1 
player_color = (0, 0, 255)  

cell_height = 57  
cell_width = 57  

positions = {}
for row in range(10):
    for col in range(10):
        x = 25 + (col if row % 2 == 0 else 9 - col) * cell_width + cell_width // 2  
        y = -20 + HEIGHT - (row + 1) * cell_height + cell_height // 2
        num = row * 10 + col + 1
        positions[num] = (x, y)

points = {2:20, 4: 25, 9: 27, 11: 31, 15:5, 18:3, 38: 60, 44:26, 46:67, 50: 68, 64: 43, 69: 34, 72: 92, 80: 82, 84: 98, 93: 75, 96: 65}

running = True
while running:
    screen.fill((255, 255, 255))  

    for event in pygame.event.get():
        if event.type == pygame.QUIT:
            running = False

        if event.type == pygame.KEYDOWN and event.key == pygame.K_SPACE:
            dice_value = random.randint(1, 6)
            new_pos = player_pos + dice_value
            if new_pos in points:
                new_pos = points[new_pos]  
            if new_pos <= 100:
                player_pos = new_pos
            print(dice_value, player_pos)
    screen.blit(board_img, (0, 0))
    pygame.draw.circle(screen, player_color, positions[player_pos], player_radius)  
    if player_pos == 100:
        break
    pygame.display.update()
pygame.quit() 
```
