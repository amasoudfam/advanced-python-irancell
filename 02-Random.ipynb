```python
def calcResult(my_choice, computer_choice):
    if my_choice == 's' and computer_choice == 'k':
        return 'computer'
    if my_choice == 'k' and computer_choice == 'g':
        return 'computer'
    if my_choice == 'g' and computer_choice == 's':
        return 'computer'
    if my_choice == computer_choice:
        return 'tie'
    return 'me'

my_score, computer_score = 0, 0
for _ in range(3):
    computer_choice = random.choice(['s', 'k', 'g'])
    my_choice = input('Enter your choice: ')
    result = calcResult(computer_choice, my_choice)
    print(my_choice, computer_choice)
    if result == 'me':
        my_score += 1
    if result == 'computer':
        computer_score += 1
if my_score > computer_score:
    print('I won!')
elif computer_score > my_score:
    print('You won!')
else:
    print('Draw')
```

```python
def calcResult(my_choice, computer_choice):
    if my_choice == 's' and computer_choice == 'k':
        return 'computer'
    if my_choice == 'k' and computer_choice == 'g':
        return 'computer'
    if my_choice == 'g' and computer_choice == 's':
        return 'computer'
    if my_choice == computer_choice:
        return 'tie'
    return 'me'

my_score, computer_score = 0, 0
while my_score < 3 and computer_score < 3:
    computer_choice = random.choice(['s', 'k', 'g'])
    my_choice = input('Enter your choice: ')
    result = calcResult(computer_choice, my_choice)
    if result == 'me':
        my_score += 1
    if result == 'computer':
        computer_score += 1
    print(my_choice, computer_choice)
    print(my_score, computer_score)
if my_score > computer_score:
    print('I won!')
elif computer_score > my_score:
    print('You won!')
else:
    print('Draw')
```

```python
import random

points = {2: 11, 10: 24, 30: 14, 40: 67, 41: 20, 70: 92, 99: 21}
current_pos = 1
while current_pos < 100:
    dice = random.randint(1, 6)
    if current_pos + dice <= 100:
        current_pos += dice
    if current_pos in points:
        print('Oooops')
        current_pos = points[current_pos]
    print(dice, current_pos)
```
