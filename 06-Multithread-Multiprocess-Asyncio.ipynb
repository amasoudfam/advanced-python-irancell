```python
def mySleep(seconds):
    start = time.time()
    while True:
        end = time.time()
        if end - start > seconds:
            break

def printHello():
    for _ in range(3):
        print('hello')
        mySleep(2)

def printBye():
    for _ in range(3):
        print('bye')
        mySleep(2)

start_time = time.time()

t1 = Thread(target=printHello)
t2 = Thread(target=printBye)

t1.start()
t2.start()

t1.join()
t2.join()
end_time = time.time()
print('time:', end_time - start_time)
print('END')
```

```python
import asyncio

async def printHello():
    for _ in range(3):
        print('hello')
        await asyncio.sleep(2)

async def printBye():
    for _ in range(3):
        print('bye')
        await asyncio.sleep(2)

await asyncio.gather(printHello(), printBye())
```

```python
import requests
from concurrent.futures import ThreadPoolExecutor
import time

NUM_REQUESTS = 500

URL = "https://httpbin.org/delay/1"
URLS = [URL] * NUM_REQUESTS

def fetch_url(url):
    response = requests.get(url)
    # print(url, len(response.content), response.status_code)

if __name__ == "__main__":
    start_time = time.time()
    with ThreadPoolExecutor(max_workers=20) as executor:
        executor.map(fetch_url, URLS)
    end_time = time.time()

    print(f"Threading - Total Time: {end_time - start_time:.2f} seconds")
```

```python
import time
import asyncio
import aiohttp

NUM_REQUESTS = 500

URL = "https://httpbin.org/delay/1"
URLS = [URL] * NUM_REQUESTS

async def fetch_url(session, url):
    async with session.get(url) as response:
        content = await response.read()
        # print(url, len(content), response.status)

async def main_async():
    async with aiohttp.ClientSession() as session:
        tasks = [asyncio.create_task(fetch_url(session, url)) for url in URLS]
        await asyncio.gather(*tasks)

if __name__ == "__main__":
    start_time = time.time()
    await main_async()
    end_time = time.time()
    print(f"Async - Total Time: {end_time - start_time:.2f} seconds")
```
